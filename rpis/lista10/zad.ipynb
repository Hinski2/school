```python
import numpy as np 
import math 

def poisson(k, lam):
    return math.exp(-lam) * lam**k / math.factorial(k)

def binom(k, n, p):
    return math.comb(n, k) * p**k * (1-p)**(n-k)

data = [
    (1, 10, 0.1, 4),
    (1, 20, 0.05, 4),
    (1, 50, 0.02, 4),
    (1, 100, 0.01, 4),
    (1, 200, 0.005, 4)
]

for lam, n, p, k in data:
    p_val = poisson(k, lam)
    b_val = binom(k, n, p)
    
    print(f"poisson = {p_val:.5f}")
    print(f"binom = {b_val:.5f}")
    print(f"diff = {(p_val - b_val):.5f}")
    print()
```

```python
import numpy as np 
import scipy.stats as stats 
import pandas as pd 

X = pd.read_csv('z1008.csv', header=None).values.flatten()

SIGMA = 3.5 
MI = 2.0

n = len(X)
print(f"n = {n}")

X_mean = np.mean(X)
print(f"X_mean = {X_mean:.5f}")

# Z ma standardardowy rozkład normalny N(0, 1)
z_stat = (X_mean - MI) / (SIGMA / np.sqrt(n))
print(f"z_stat = {z_stat:.5f}")

f_z = stats.norm.cdf(z_stat)
print(f"f_z = {f_z:.5f}")
```

# Zad 9

```python
import numpy as np 
import scipy.stats as stats
import pandas as pd

X = pd.read_csv('z1008.csv', header=None).values.flatten()
MI = 2.0 

n = len(X)
STOPNIE_SWOBODY = n - 1
print(f"n = {n}")

X_mean = np.mean(X)
print(f"X_mean = {X_mean:.5f}")

s_mle = np.std(X, ddof=0)
print(f"s_mle = {s_mle:.5f}")

# statystyka T-studenta
t_stat = ((X_mean - MI) / s_mle) * np.sqrt(STOPNIE_SWOBODY)
print(f"t_stat = {t_stat:.5f}")

f_t = stats.t.cdf(t_stat, df=STOPNIE_SWOBODY)
print(f"f_t = {f_t:.5f}")
```
